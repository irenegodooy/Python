#### 9.- Dada la lista _valores_, convierte cada número a string, pero con las siguientes reglas:
- Si es par: "par:<num>"
- Si es múltiplo de 4: "mult3:<num>"
- Si cumple ambas condiciones, concatenar: "par-mult3:<num>"
#### Devuelve la lista final.

```python
valores = [10,21,33,42,55]
```

El enunciado dice múltiplos de **4**, pero usa la etiqueta **mult3**. Se incluyen ambas versiones. Todo múltiplo de 4 es par: en la versión literal nunca aparece la etiqueta mult3 sola. Los números sin condición se convierten a texto sin prefijo.

```python
# Versión literal: múltiplos de 4, con la etiqueta "mult3" del enunciado.
resultado = []
for numero in valores:
    if numero % 2 == 0 and numero % 4 == 0:
        resultado.append("par-mult3:" + str(numero))
    elif numero % 2 == 0:
        resultado.append("par:" + str(numero))
    elif numero % 4 == 0:
        resultado.append("mult3:" + str(numero))
    else:
        resultado.append(str(numero))
print("Versión con múltiplos de 4:", resultado)

# Alternativa: si el enunciado quería decir múltiplos de 3.
resultado_alternativo = []
for numero in valores:
    if numero % 2 == 0 and numero % 3 == 0:
        resultado_alternativo.append("par-mult3:" + str(numero))
    elif numero % 2 == 0:
        resultado_alternativo.append("par:" + str(numero))
    elif numero % 3 == 0:
        resultado_alternativo.append("mult3:" + str(numero))
    else:
        resultado_alternativo.append(str(numero))
print("Versión con múltiplos de 3:", resultado_alternativo)
```

#### 10.- Dado el dicionario _datos_, pon la primera letra de "nombre" en mayúsculas. Calcula la suma total de "puntos" y guarda el total en una nueva clave "total".
```python
datos = [
    {"id":1,"nombre":"ana","puntos":(10,20,30)},
    {"id":2,"nombre":"luis","puntos":(5,15)},
    {"id":3,"nombre":"marta","puntos":(100,50,25)}
]
```

#### 12.- Dado el diccionario _estudiantes_, crea un nuevo diccionario con la siguiente estructura:
- La clave debe ser el nombre en mayúsculas
- El valor debe ser otro diccionario con:
    - "promedio" como promedio de notas
    - "aprobado" como True si el promedio es >= 20, de lo contrario es False
    - "max_min" como tupla con (nota_max,nota_min)

```python
estudiantes = {
    "s1" : {"nombre":"ana","notas":[10,20,30]},
    "s2" : {"nombre":"luis","notas":[15,5]},
    "s3" : {"nombre":"marta","notas":[50,100,25]},
    "s4" : {"nombre":"pepe","notas":[40]}
}
```

```python
resumen = {}
for estudiante in estudiantes.values():
    nombre = estudiante["nombre"].upper()
    notas = estudiante["notas"]
    promedio = sum(notas) / len(notas)
    resumen[nombre] = {
        "promedio": promedio,
        "aprobado": promedio >= 20,
        "max_min": (max(notas), min(notas))
    }
print(resumen)
```

#### 13.- Crea un archivo llamado _notas_alumnos.txt_ con este contenido:
#### Ana,10,20,30
#### Luis,5,15
#### Marta,100,50,25
#### Convierte cada línea en un diccionario con "nombre" capitalizado, "promedio" y "aprobado" si promedio >= 20.
#### Guarda la lista final en un archivo resultado.json

```python
import json

with open("notas_alumnos.txt", "w", encoding="utf-8") as archivo:
    archivo.write("Ana,10,20,30\nLuis,5,15\nMarta,100,50,25\n")

resultado_alumnos = []
with open("notas_alumnos.txt", "r", encoding="utf-8") as archivo:
    for linea in archivo:
        partes = linea.strip().split(",")
        notas = []
        for numero in partes[1:]:
            notas.append(int(numero))
        promedio = sum(notas) / len(notas)
        alumno = {
            "nombre": partes[0].capitalize(),
            "promedio": promedio,
            "aprobado": promedio >= 20
        }
        resultado_alumnos.append(alumno)

with open("resultado.json", "w", encoding="utf-8") as archivo:
    json.dump(resultado_alumnos, archivo, ensure_ascii=False, indent=4)
print(resultado_alumnos)
```

El ejercicio menciona datos con roles, pero no los define; los datos del ejercicio 10 contienen puntos. Aquí se crea una lista con los usuarios y roles del ejemplo del ejercicio 14.

```python
import json
import csv

# El enunciado no aporta datos con roles: usamos los de su ejemplo.
datos = [
    {"id": 1, "nombre": "ana", "roles": ["user", "admin"]},
    {"id": 2, "nombre": "luis", "roles": ["user"]},
    {"id": 3, "nombre": "marta", "roles": ["moderator", "user"]}
]
with open("usuarios.json", "w", encoding="utf-8") as archivo:
    json.dump(datos, archivo, ensure_ascii=False, indent=4)

with open("usuarios.json", "r", encoding="utf-8") as archivo:
    usuarios_leidos = json.load(archivo)

for usuario in usuarios_leidos:
    usuario["nombre"] = usuario["nombre"].upper()

with open("usuarios.csv", "w", encoding="utf-8", newline="") as archivo:
    escritor = csv.writer(archivo)
    escritor.writerow(["id", "nombre", "roles"])
    for usuario in usuarios_leidos:
        escritor.writerow([usuario["id"], usuario["nombre"], "|".join(usuario["roles"])])
print(usuarios_leidos)
```

#### 15.- Con el archivo _fragmento-libro.txt_, ábrelo, cuenta la frecuencia de cada palabra (ignorando mayúsculas/minúsculas y signos). Devuelve un archivo llamado _frecuencias.json_ con las 5 palabras más comunes. 

```python
import json

with open("fragmento-libro.txt", "r", encoding="utf-8") as archivo:
    contenido = archivo.read().lower()

# Conservamos las letras (incluidas tildes y ñ) y separamos los signos.
texto_limpio = ""
for caracter in contenido:
    if caracter.isalpha():
        texto_limpio = texto_limpio + caracter
    else:
        texto_limpio = texto_limpio + " "

frecuencias_libro = {}
for palabra in texto_limpio.split():
    if palabra not in frecuencias_libro:
        frecuencias_libro[palabra] = 0
    frecuencias_libro[palabra] = frecuencias_libro[palabra] + 1

def obtener_frecuencia_libro(elemento):
    return elemento[1]

ordenadas = sorted(frecuencias_libro.items(), key=obtener_frecuencia_libro, reverse=True)
cinco_comunes = dict(ordenadas[:5])
with open("frecuencias.json", "w", encoding="utf-8") as archivo:
    json.dump(cinco_comunes, archivo, ensure_ascii=False, indent=4)
print(cinco_comunes)
```

#### 16.- Crea tres archivos CSV (_ventas1.csv_, _ventas2.csv_, _ventas3.csv_ con la misma información:
- producto,cantidad
- A,10
- B,5
#### Combínalos en un único archivo que se llema _total_ventas.csv_. Suma las cantidades por producto (si el producto se repite).


```python
import csv

archivos = ["ventas1.csv", "ventas2.csv", "ventas3.csv"]
for nombre_archivo in archivos:
    with open(nombre_archivo, "w", encoding="utf-8", newline="") as archivo:
        escritor = csv.writer(archivo)
        escritor.writerow(["producto", "cantidad"])
        escritor.writerow(["A", 10])
        escritor.writerow(["B", 5])

totales = {}
for nombre_archivo in archivos:
    with open(nombre_archivo, "r", encoding="utf-8", newline="") as archivo:
        lector = csv.DictReader(archivo)
        for fila in lector:
            producto = fila["producto"]
            cantidad = int(fila["cantidad"])
            if producto not in totales:
                totales[producto] = 0
            totales[producto] = totales[producto] + cantidad

with open("total_ventas.csv", "w", encoding="utf-8", newline="") as archivo:
    escritor = csv.writer(archivo)
    escritor.writerow(["producto", "cantidad"])
    for producto in totales:
        escritor.writerow([producto, totales[producto]])
print(totales)
```

#### 17.- Crea un archivo _alumnos.json_ con el contenido siguiente:
- [{"nombre": "ana", "notas": [10, 20, 30]},{"nombre": "luis", "notas": [5, 15]},{"nombre": "marta", "notas": [100, 50, 25]}]
#### Lee el archivo y genera un diccionario donde la clave sea el nombre y el valor tenga el siguiente formato:
- {"promedio":X,"max":Y,"min":Z,"aprobado":True/False}


Como no se indica el umbral de aprobado, se utiliza promedio >= 20, igual que en los ejercicios 12 y 13.

```python
import json

alumnos = [
    {"nombre": "ana", "notas": [10, 20, 30]},
    {"nombre": "luis", "notas": [5, 15]},
    {"nombre": "marta", "notas": [100, 50, 25]}
]
with open("alumnos.json", "w", encoding="utf-8") as archivo:
    json.dump(alumnos, archivo, ensure_ascii=False, indent=4)

with open("alumnos.json", "r", encoding="utf-8") as archivo:
    alumnos_leidos = json.load(archivo)

resumen_alumnos = {}
for alumno in alumnos_leidos:
    notas = alumno["notas"]
    promedio = sum(notas) / len(notas)
    resumen_alumnos[alumno["nombre"]] = {
        "promedio": promedio,
        "max": max(notas),
        "min": min(notas),
        "aprobado": promedio >= 20
    }
print(resumen_alumnos)
```
